```python
files = [
    'load-data-total-28468.csv',
    'load-data-total-76690.csv',
    'load-data-total-120749.csv',
    'load-data-total-246148.csv',
    'load-data-total-283599.csv',
    'load-data-total-380447.csv',
    'load-data-total-417488.csv', 
    'load-data-total-433876.csv',
    'load-data-total-544473.csv'
]
```

```python
import pandas as pd

for file_name in files:

    df = pd.read_csv(file_name)

    hourly_data = [0] * 8760
    for i, v in enumerate(df['out.electricity.total.energy_consumption'].values):
        hourly_data[int(i / 4)] += v

    idx = range(0, 8760)
    df_hrly = pd.DataFrame.from_dict({'time' : idx, 'kW' : hourly_data})
    df_hrly.to_csv(file_name.split(".")[0] + '_hourly.csv')
```
